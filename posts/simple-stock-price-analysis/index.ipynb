Let's explore how moving averages can help us understand stock price trends. We'll generate some synthetic stock data and analyze it using different moving average periods.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate synthetic stock data
np.random.seed(42)
days = 200
price = 100
prices = [price]

for _ in range(days - 1):
    change_percent = np.random.normal(0.001, 0.02)
    price = price * (1 + change_percent)
    prices.append(price)

# Create DataFrame
df = pd.DataFrame({
    'Date': pd.date_range(start='2023-01-01', periods=days),
    'Price': prices
})
df.set_index('Date', inplace=True)
```

```python
# Calculate moving averages
df['MA20'] = df['Price'].rolling(window=20).mean()
df['MA50'] = df['Price'].rolling(window=50).mean()

# Plot the data
plt.figure(figsize=(12, 6))
plt.plot(df.index, df['Price'], label='Price', alpha=0.7)
plt.plot(df.index, df['MA20'], label='20-day MA', linewidth=2)
plt.plot(df.index, df['MA50'], label='50-day MA', linewidth=2)
plt.title('Stock Price with Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```

The 20-day moving average (MA) responds more quickly to price changes, while the 50-day MA shows longer-term trends. Traders often look for crossovers between these lines as potential trading signals.
