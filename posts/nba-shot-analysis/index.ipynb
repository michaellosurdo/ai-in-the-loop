In this post, we'll explore a simple analysis of NBA shot patterns using Python. We'll create some synthetic data and analyze how shooting percentage varies with distance.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic shot data
n_shots = 1000
distances = np.random.normal(15, 5, n_shots)  # Most shots between 10-20 feet
# Shooting percentage decreases with distance
probabilities = 0.6 - 0.02 * distances + np.random.normal(0, 0.1, n_shots)
made_shots = np.random.binomial(1, np.clip(probabilities, 0, 1))

# Create DataFrame
shot_data = pd.DataFrame({
    'distance': distances,
    'made': made_shots
})
```

```python
# Calculate shooting percentage by distance range
shot_data['distance_range'] = pd.cut(shot_data['distance'], bins=10)
shooting_pct = shot_data.groupby('distance_range')['made'].mean()

# Create visualization
plt.figure(figsize=(10, 6))
shooting_pct.plot(kind='bar')
plt.title('Shooting Percentage by Distance')
plt.xlabel('Distance Range (feet)')
plt.ylabel('Shooting Percentage')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

As we can see from the visualization, there's a clear negative correlation between shot distance and shooting percentage, which matches what we would expect in real basketball games.
